```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('factbook.db')
```

```python
query1 = """
SELECT cities.name AS city, cities.population
FROM cities
JOIN facts ON cities.facts_id = facts.id
WHERE facts.area > 5000;
"""
pd.read_sql_query(query1, conn)
```

```python
query2 = """
SELECT cities.name AS capital_city
FROM cities
JOIN facts ON cities.facts_id = facts.id
WHERE cities.capital = 1
  AND facts.death_rate > 2
  AND facts.birth_rate < 20;
"""
pd.read_sql_query(query2, conn)
```

```python
query3 = """
SELECT cities.name AS city, facts.population_growth
FROM cities
JOIN facts ON cities.facts_id = facts.id
WHERE facts.population_growth BETWEEN 1 AND 5
ORDER BY facts.population_growth DESC;
"""
pd.read_sql_query(query3, conn)
```

```python
query4 = """
SELECT facts.name AS country, cities.name AS city
FROM facts
LEFT JOIN cities ON facts.id = cities.facts_id
WHERE facts.name LIKE 'A%'
  AND cities.id IS NULL;
"""
pd.read_sql_query(query4, conn)
```

```python
query5 = """
SELECT cities.name AS capital_city, facts.death_rate
FROM cities
LEFT JOIN facts ON cities.facts_id = facts.id
WHERE cities.capital = 1;
"""
pd.read_sql_query(query5, conn)
```
