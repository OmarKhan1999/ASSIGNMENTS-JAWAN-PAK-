```python
# Question 4

course_a = {"John", "Alice", "Bob", "David"}
course_b = {"Alice", "Eve", "Charlie", "David"}

print('Students who are in Course A but not in Course B =',course_a.difference(course_b))
print('Students who are only in one of the two courses =',course_a.symmetric_difference(course_b))
```

```python
# Question 6

student_details = {
    "Ali": {"age": 20, "grade": "A", "city": "Karachi"},
    "Sara": {"age": 21, "grade": "B", "city": "Lahore"},
    "Omar": {"age": 19, "grade": "A", "city": "Islamabad"},
    "Hina": {"age": 22, "grade": "C", "city": "Faisalabad"}
}

# Adding Zara
student_details.update({
    "Zara": {
        "age": 20,
        "grade": "B",
        "city": "Multan"
    }
})

# Removing Omar
student_details.pop("Omar")

print(student_details)
```

```python
# Question 7

inventory = {
    "The Great Gatsby": {"price": 10.99, "quantity": 5},
    "Rich dad poor dad": {"price": 8.99, "quantity": 8},
    "Harry Potter": {"price": 12.50, "quantity": 2}
}

# Adding a new book
inventory.update({
    "Atomic Habits": {"price": 15.99, "quantity": 6}
})

# all book titles
print(inventory.keys())

# Harry Potter is out of stock
inventory["Harry Potter"]["quantity"] = 0

# Remove it
inventory.pop("Harry Potter")

print(inventory)
```

```python
# Question 10

A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7}

print('Common Elements=',A.intersection(B))
print('Only in Set A =',A.difference(B))
print('Only in Set B =',B.difference(A))
```
