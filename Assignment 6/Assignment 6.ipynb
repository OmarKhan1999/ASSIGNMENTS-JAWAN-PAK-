```python
# Q1

secret_word = "python"
guess = ""

while guess != secret_word:
    guess = input("Guess the word: ")

print("🎉 You guessed it!")
```

```python
# Q2

users = ["Ali", "Sara", "John", "Zara", "Mike"]
count = 0

for user in users:
    if count == 3:
        print("Limit reached. Stop sending emails.")
        break
    print(f"Email sent to {user}")
    count += 1
```

```python
# Q3

temps = [22, 25, 30, 45, 28]
critical = 40

for temp in temps:
    if temp >= critical:
        print("Critical temperature reached! Stop monitoring.")
        break
    print(f"Temperature: {temp}")
```

```python
# Q4

cart = [100, 250, 50, 300]
total = 0

for price in cart:
    total += price

print(f"Total Price: {total}")
```

```python
# Q5

customers = [
    {"name": "Ali", "paid": True},
    {"name": "Sara", "paid": False},
    {"name": "John", "paid": True}
]

for customer in customers:
    if not customer["paid"]:
        continue
    print(f"Access granted to: {customer['name']}")
```

```python
# Q6

users = ["Ali", "Sara", "John"]

for user in users:
    print(f"Hello {user}, your subscription is active!")
```

```python
# Q7

emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]
domains = []

for email in emails:
    domain = email.split("@")[1]
    domains.append(domain)

print("domains:", domains)
```

```python
# Q8

tasks = [
    {"task": "Pay bills", "status": "done"},
    {"task": "Submit report", "status": "pending"},
    {"task": "Clean house", "status": "pending"}
]

for t in tasks:
    if t["status"] == "pending":
        print(f"Overdue: {t['task']}")
```

```python
# Q9

while True:
    num = int(input("Enter a number (negative to stop): "))
    if num < 0:
        print("Negative number entered. Loop stopped!")
        break
    print(f"You entered: {num}")
```

```python
# Q10

days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

for day in days:
    if day == "Saturday" or day == "Sunday":
        continue
    print(f"Workday: {day}")
```
