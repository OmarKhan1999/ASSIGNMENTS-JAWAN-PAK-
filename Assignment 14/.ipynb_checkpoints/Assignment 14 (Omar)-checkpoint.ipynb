```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("database.sqlite")
```

```python
query1 = """
SELECT JobTitle,
       AVG(TotalPayBenefits) AS AverageTotalPayBenefits
FROM Salaries
WHERE JobTitle NOT IN ('Not Provided', 'Not provided')
GROUP BY JobTitle
ORDER BY AverageTotalPayBenefits DESC
"""

df1 = pd.read_sql_query(query1, conn)
df1
```

```python
query2 = """
SELECT EmployeeName, JobTitle, OvertimePay
FROM Salaries
WHERE CAST(OvertimePay AS REAL) > 50000
ORDER BY OvertimePay DESC
"""

df2 = pd.read_sql_query(query2, conn)
df2
```

```python
query3 = """
SELECT COUNT(DISTINCT EmployeeName) AS UniqueEmployees
FROM Salaries
WHERE EmployeeName NOT IN ('Not Provided', 'Not provided')
"""

df3 = pd.read_sql_query(query3, conn)
df3
```

```python
query4 = """
SELECT EmployeeName, JobTitle, TotalPayBenefits
FROM Salaries
WHERE EmployeeName NOT IN ('Not Provided', 'Not provided')
ORDER BY TotalPayBenefits DESC
LIMIT 10
"""

df4 = pd.read_sql_query(query4, conn)
df4
```

```python
query5 = """
SELECT *
FROM Salaries
ORDER BY TotalPayBenefits DESC
"""

df5 = pd.read_sql_query(query5, conn)
df5
```

```python
query5 = """
SELECT EmployeeName, TotalPayBenefits
FROM Salaries
ORDER BY TotalPayBenefits DESC
"""

df5 = pd.read_sql_query(query5, conn)
df5
```
