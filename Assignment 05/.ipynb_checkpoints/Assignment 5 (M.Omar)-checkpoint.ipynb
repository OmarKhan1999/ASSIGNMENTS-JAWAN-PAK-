```python
# Question 1

age = int(input("Enter your age: "))

if age >= 0 and age <= 12:
    print("Child")
elif age >= 13 and age <= 19:
    print("Teenager")
elif age >= 20 and age <= 64:
    print("Adult")
elif age >= 65:
    print("Senior")
else:
    print("Invalid Age")
```

```python
# Question 2

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
num3 = int(input("Enter third number: "))

largest = max(num1, num2, num3)
print("Largest number is:", largest)

if num1 == num2 == num3:
    print("All numbers are equal")
elif num1 == num2 or num1 == num3 or num2 == num3:
    print("Some numbers are equal")
else:
    print("All numbers are different")
```

```python
# Question 3

my_tuple = (5, 10, 15, 20, 25)

num = int(input("Enter a number: "))

if num in my_tuple:
    print("Number exists in the tuple.")
else:
    print("Number does not exist in the tuple.")
```

```python
# Question 4

title = input("Enter book title: ")
author = input("Enter author name: ")
year = int(input("Enter year of publication: "))
price = float(input("Enter price: "))

book = (title, author, year, price)

print("Title:", book[0])
print("Author:", book[1])
print("Year of Publication:", book[2])
print("Price: $", book[3])
```

```python
# Question 5

subjects = ["math", "english", "science", "art", "computer"]

subject = input("Enter the subject you are interested in: ")

if subject.lower() in subjects:
    print(subject, "is available.")
else:
    print(subject, "is not available.")
```

```python
# Question 6

number = int(input("Enter a number: "))

if number > 0:
    print("The number is Positive")
elif number < 0:
    print("The number is Negative")
else:
    print("The number is Zero")
```

```python
# Question 7

username = input("Enter username: ")
password = input("Enter password: ")

if username == "admin" and password == "admin@123":
    print("Login Successful")
elif username != "admin" and password == "admin@123":
    print("Invalid Username")
elif username == "admin" and password != "admin@123":
    print("Invalid Password")
else:
    print("Invalid Username and Password")
```

```python
# Question 8

age = int(input("Enter your age: "))
credit_score = int(input("Enter your credit score: "))
income = int(input("Enter your annual income: "))

if age < 18:
    print("You are not eligible for the loan due to age.")
elif credit_score < 650:
    print("You are not eligible for the loan due to low credit score.")
elif credit_score >= 700 and income >= 25000:
    print("You are eligible for the loan!")
elif income >= 30000:
    print("You are eligible for the loan!")
else:
    print("You are not eligible for the loan due to low income.")
```

```python
# Question 9

letter = input("Enter a single letter: ")

if len(letter) == 1 and letter.isalpha():
    if letter.lower() in "aeiou":
        print(letter, "is a Vowel")
    else:
        print(letter, "is a Consonant")
else:
    print("Please enter valid alphabet")
```
