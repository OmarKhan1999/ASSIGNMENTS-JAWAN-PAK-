```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings 
import pandas as pd
df = pd.read_csv("titanic_dataset.csv")
```

```python
#Q1

survival_rate = df.groupby("Embarked")["Survived"].mean()

sns.barplot(x=survival_rate.index, y=survival_rate.values)

plt.title("Survival Rate by Embarked Port")
plt.xlabel("Embarked")
plt.ylabel("Average of Survived")
plt.show()

print(survival_rate)
```

```python
print(survival_rate.idxmax())
```

```python
#Q2

df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
```

```python
sns.countplot(x="FamilySize", data=df)

plt.title("Number of Passengers by Family Size")
plt.xlabel("FamilySize")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
print(df["FamilySize"].value_counts().idxmax())
```

```python
#Q3

sns.boxplot(x="Pclass", y="Fare", hue="Sex", data=df)

plt.title("Fare by Passenger Class and Sex")
plt.xlabel("Pclass")
plt.ylabel("Fare")
plt.show()
```

```python
print(df.loc[df["Fare"].idxmax(), ["Pclass", "Sex", "Fare"]])
```

```python
#Q4

survival_family = df.groupby("FamilySize")["Survived"].mean()

plt.plot(survival_family.index, survival_family.values, marker="o")

plt.title("Survival Rate by Family Size")
plt.xlabel("FamilySize")
plt.ylabel("Survival Rate")
plt.show()
```

```python
print(survival_family.idxmax())
```

```python
#Q5

survival_count = pd.crosstab(
    [df["Pclass"], df["Sex"]],
    df["Survived"]
)
```

```python
survival_count.columns = ["Not Survived", "Survived"]
```

```python
survival_count.plot(kind="bar", stacked=True)

plt.title("Survival Comparison by Passenger Class and Gender")
plt.xlabel("Pclass and Sex")
plt.ylabel("Number of Passengers")
plt.legend(title="Survival")
plt.show()
```
