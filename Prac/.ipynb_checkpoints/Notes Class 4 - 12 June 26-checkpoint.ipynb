## Indexing of Lists


```python
salaries=['100K', '200K', '300k', '400k', '500k', '600k','700k']
```

```python
salaries
```

```python
# Positive Intexing

salaries[3]
```

```python
# Negative Indexing

salaries[-3]
```

```python
# Length of all items in list

len(salaries)
```

```python
# To find the location of specific items in a list

salaries.index('500k')
```

```python
# Nested List

bio = ['Ali', 25, salaries, 40.2]
```

```python
bio
```

```python
# Normal Indexing

bio[2]
```

```python
# Nested Indexing Calling values indside a list in a list

bio[2][3]
```

# Slicing


```python
salaries
```

```python
# Slicing 2 Represents Starting value,
# 7 Represents Ending Value and 
# 1 Represents the intervention between the nuumbers (no. of skips)

salaries[2:7:1]
```

```python
# Slicing 2 Represents Starting value,
# 7 Represents Ending Value and 
# 2 Represents the intervention between the nuumbers (no. of skips)

salaries[2:7:2]
```

```python
# Normal Slicing with 1st value as -1 position of the needed value (Starting from Zero) and 2nd value as exact value needed

salaries [3:6]
```

```python
# Negative Slicing with 1st value as exact value needed and 2nd value as -1 position of the needed value (Starting from Zero)

salaries[-4:-1]
```

```python
# to sort the items in a list (sort) for assending order

# variable.sort() FOR ASSENDING
# variable.sort(reverse=True) FOR DECENDING

# FOR EXAMPLE:

test=[2,5,6,9,3,4,0]

test.sort() # Assending
print(test)

test.sort(reverse=True) # Desending
print(test)
```

```python
# There is no function to replace a specific item in a list so we do it manually.
# this is how we replace a slice of items with new one
# there's no limit to items that can replace the given range (3 items can be replaced by 2 items as well as 200)
# so make sure to add items for the range selected to be replaced

# NEW LIST
prac = [1,2,3,4,5,6,7]

# RANGE SELECTED TO BE REPLACED
print(prac [2:5])

# RANGE REPLACED WITH NEW VALUES
prac[2:5] = [20,30,40]

# FINAL PRINT
print(prac)


# TO REPLACE A SPECIFIC ITEM WITH MULTIPLE IETMS

print(prac[5])

prac[5] = ['a','b','c','d']
# REPLACES THE ITEM WITH A LIST (makes its a tupple)

prac
```
