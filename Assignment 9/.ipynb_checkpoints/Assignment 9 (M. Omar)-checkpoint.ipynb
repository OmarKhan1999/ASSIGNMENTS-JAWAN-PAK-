```python
import csv

# Open the original CSV file
with open("students.csv", "r") as file:

    reader = csv.DictReader(file)

    # Get the existing column names
    fieldnames = reader.fieldnames + ["Grade"]

    # Create a new CSV file
    with open("students_with_grades.csv", "w", newline="") as new_file:

        writer = csv.DictWriter(new_file, fieldnames=fieldnames)

        # Write the headings
        writer.writeheader()

        # Read each students
        for student in reader:

            percentage = float(student["Percentage"])

            # Give grade according to percentage
            if percentage > 80:
                grade = "A"
            elif percentage > 70:
                grade = "B"
            elif percentage > 60:
                grade = "C"
            elif percentage > 50:
                grade = "D"
            else:
                grade = "F"

            student["Grade"] = grade

            # Write student data
            writer.writerow(student)

print("Grades added successfully!")
```
