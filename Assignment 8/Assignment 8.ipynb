```python
# Question 1

class Employee:

    def __init__(self, name, number, address, salary, post, year_of_joining, education, year_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    def describe_employee(self):
        print("Name:", self.name)
        print("Number:", self.number)
        print("Address:", self.address)
        print("Salary:", self.salary)
        print("Post:", self.post)
        print("Year of Joining:", self.year_of_joining)
        print("Education:", self.education)
        print("Year of Experience:", self.year_of_experience)
        print("----------------------------")

    def update_year_of_experience(self):
        self.year_of_experience += 1

    def increase_salary(self, amount):
        self.salary += amount


# Employee 101
employee101 = Employee(
    "Ali",
    101,
    "Karachi",
    50000,
    "Accountant",
    2022,
    "BBA",
    3
)

# Print four attributes
print(employee101.name)
print(employee101.number)
print(employee101.salary)
print(employee101.post)

# Call methods
employee101.describe_employee()

employee101.increase_salary(5000)
employee101.update_year_of_experience()

employee101.describe_employee()


# Create 10 Employees

employees = [
    Employee("Ali",101,"Karachi",50000,"Accountant",2022,"BBA",3),
    Employee("Ahmed",102,"Lahore",45000,"Cashier",2021,"B.Com",4),
    Employee("Sara",103,"Islamabad",60000,"Manager",2019,"MBA",6),
    Employee("Ayesha",104,"Multan",55000,"HR",2020,"MBA",5),
    Employee("Usman",105,"Peshawar",48000,"Sales",2023,"BBA",2),
    Employee("Hina",106,"Quetta",70000,"Director",2015,"MBA",10),
    Employee("Bilal",107,"Faisalabad",52000,"Officer",2021,"BBA",4),
    Employee("Omar",108,"Hyderabad",47000,"Assistant",2022,"BSCS",3),
    Employee("Zara",109,"Sialkot",65000,"Supervisor",2018,"MBA",7),
    Employee("Hamza",110,"Rawalpindi",53000,"Coordinator",2020,"BBA",5)
]

# Display all employees
for emp in employees:
    emp.describe_employee()

# Update first employee
employees[0].increase_salary(3000)
employees[0].update_year_of_experience()

print("After Update")
employees[0].describe_employee()
```
